### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# Conteo de duplicados.

dup_first = hoteles.duplicated(keep='first').sum()
dup_last = hoteles.duplicated(keep='last').sum()
dup_false = hoteles.duplicated(keep=False).sum()

print(f"Duplicados keep='first': {dup_first}")

print(f"Duplicados keep='last': {dup_last}")

print(f"Duplicados keep=False: {dup_false}")
```

*   Duplicados con keep=False: **40,165**


El hecho de que los valores de keep='first' y keep='last' sean iguales indica que la cantidad de registros repetidos es la misma independientemente de si se conserva la primera o la última aparición de cada grupo de filas idénticas.

El resultado de 40,165 filas con keep=False significa que ese número de registros pertenece a algún grupo de duplicados. Es decir, aproximadamente el 33.6% de las observaciones del conjunto de datos forman parte de grupos de filas completamente idénticas:

(40165/119390)*100= 33.6%

---



La diferencia entre 40,165 y 31,994 se debe a que keep=False cuenta todas las filas involucradas en duplicados, mientras que keep='first' y keep='last' excluyen una ocurrencia de cada grupo.

Debido a que no existe un identificador único de reserva, no puede afirmarse  que se trate de errores o duplicados reales. Es posible que algunas reservas distintas tengan exactamente los mismos valores en las 32 variables registradas.

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Filas sobrantes
filas_sobrantes = hoteles.duplicated(keep='first').sum()

# Porcentaje
porcentaje_sobrantes = (filas_sobrantes / len(hoteles)) * 100

# Grupos repetidos
grupos_repetidos = (
    hoteles.groupby(list(hoteles.columns))
           .size()
           .gt(1)
           .sum()
)

print(f"Porcentaje de filas sobrantes: {porcentaje_sobrantes:.2f}%")
print(f"Grupos repetidos: {grupos_repetidos}")
```

El **26.80%** de las filas son registros repetidos respecto a una observación idéntica previa. En términos prácticos, esto significa que 31,994 de las 119,390 filas podrían eliminarse si se conserva una sola ocurrencia de cada registro duplicado.

Se encontraron **17 grupos** repetidos, por lo que solo 17 combinaciones de valores distintas aparecen más de una vez. Esto indica que los duplicados no están distribuidos en muchos patrones diferentes.


### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Copia
hoteles_dup = hoteles.copy()

# Booleana de duplicados
hoteles_dup['es_duplicado'] = hoteles_dup.duplicated(keep='first')

# Datos por Hotel
resumen = hoteles_dup.groupby('hotel').agg(
    total_reservas=('hotel', 'size'),
    filas_sobrantes=('es_duplicado', 'sum')
)

resumen['porcentaje_sobrantes'] = (
    resumen['filas_sobrantes'] / resumen['total_reservas'] * 100
)

print(resumen)
```

El City Hotel tiene **25,902** filas sobrantes 32.65%, mientras que el Resort Hotel tiene 6,092 filas sobrantes 15.21%.
Esto indica que City Hotel tiene mas registros repedidos.

## Actividad 2. Inspección y criterio de duplicado (25 puntos)

### 2.1 Ver los grupos repetidos (6 pts)

Filtra las filas involucradas en duplicados (`keep=False`), **ordénalas** con `sort_values()` para que las copias queden juntas y muestra las primeras 12 filas con estas columnas:

`hotel`, `lead_time`, `arrival_date_year`, `arrival_date_month`, `arrival_date_day_of_month`, `adults`, `country`, `market_segment`, `deposit_type`, `adr`, `reservation_status`.

Elige tú las columnas de ordenamiento y explica tu elección en un comentario.

```python

duplicados = hoteles[hoteles.duplicated(keep=False)]

# Ordenamiento elegido:
# Se ordena por las variables que describen la reserva para que las filas idénticas queden juntas y
# sea más fácil identificar grupo duplicado.
duplicados = duplicados.sort_values(
    by=[
        'hotel',
        'lead_time',
        'arrival_date_year',
        'arrival_date_month',
        'arrival_date_day_of_month',
        'adults',
        'country',
        'market_segment',
        'deposit_type',
        'adr',
        'reservation_status'
    ]
)

# Mostrar columnas
columnas = [
    'hotel',
    'lead_time',
    'arrival_date_year',
    'arrival_date_month',
    'arrival_date_day_of_month',
    'adults',
    'country',
    'market_segment',
    'deposit_type',
    'adr',
    'reservation_status'
]

print(duplicados[columnas].head(12))
```

Se realizo el orden por hotel, fecha de llegada, lead_time, número de adultos, país, segmento de mercado, tipo de depósito, tarifa (adr) y estado de la reserva, porque estas variables describen las características de una reserva. Al utilizar estas columnas en sort_values(), las filas con valores idénticos quedan agrupadas una junto a la otra, haciendo mas facil identificar las repetidas.

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
hoteles_limpio = hoteles.drop_duplicates(keep='first').reset_index(drop=True)

# Cantidad de filas
filas_antes = len(hoteles)
filas_despues = len(hoteles_limpio)
filas_eliminadas = filas_antes - filas_despues

print(f"Filas antes de eliminar duplicados: {filas_antes}")
print(f"Filas después de eliminar duplicados: {filas_despues}")
print(f"Filas eliminadas: {filas_eliminadas}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
resumen_original = (
    hoteles.groupby('hotel')
    .agg(
        reservas=('hotel', 'size'),
        tasa_cancelacion=('is_canceled', 'mean'),
        adr_promedio=('adr', 'mean')
    )
    .reset_index()
)

# sin duplicados
resumen_limpio = (
    hoteles_limpio.groupby('hotel')
    .agg(
        reservas=('hotel', 'size'),
        tasa_cancelacion=('is_canceled', 'mean'),
        adr_promedio=('adr', 'mean')
    )
    .reset_index()
)

print("Resumen original")
print(resumen_original)

print("\nResumen limpio")
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python

resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

comparacion = pd.concat(
    [resumen_original, resumen_limpio],
    ignore_index=True
)

comparacion = comparacion.sort_values(
    by=['hotel', 'version']
).reset_index(drop=True)

print(comparacion)
```
